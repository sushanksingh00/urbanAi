```python
import pickle
from pathlib import Path
import pandas as pd

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/urbanAi/data/processed/METR-LA"
)
ADJ_PATH = PROCESSED_DIR / "adj_METR-LA.pkl"

# Indices identified by Notebook 11
important_indices = [82, 76, 60, 81, 43]

with open(ADJ_PATH, "rb") as f:
    adj_data = pickle.load(f, encoding="latin1")

sensor_ids = adj_data[0]
sensor_id_to_index = adj_data[1]

print("Number of sensor IDs:", len(sensor_ids))
print("\nTop sensitive sensors:")

for idx in important_indices:
    print(f"Index {idx}: Sensor ID {sensor_ids[idx]}")

# Save the exact mapping for your research records
mapping_df = pd.DataFrame({
    "sensor_index": important_indices,
    "sensor_id": [sensor_ids[i] for i in important_indices],
})
mapping_df.to_csv(
    "/content/experiments/explainability/top_sensitive_sensor_ids.csv",
    index=False,
)
display(mapping_df)
```

```python
import pandas as pd
from pathlib import Path

url = (
    "https://raw.githubusercontent.com/liyaguang/"
    "DCRNN/master/data/sensor_graph/graph_sensor_locations.csv"
)

locations = pd.read_csv(url)

top_sensors = pd.DataFrame({
    "sensor_index": [82, 76, 60, 81, 43],
    "sensor_id": ["769430", "717499", "769444", "769431", "760024"]
})

locations["sensor_id"] = locations["sensor_id"].astype(str)
top_sensors["sensor_id"] = top_sensors["sensor_id"].astype(str)

result = top_sensors.merge(
    locations,
    on="sensor_id",
    how="left",
    validate="one_to_one",
    indicator=True
)

display(result)

print("Matched:", (result["_merge"] == "both").sum(), "of", len(result))

# Save the verified mapping
output_dir = Path("/content/experiments/explainability")
output_dir.mkdir(parents=True, exist_ok=True)

result.to_csv(
    output_dir / "top_sensitive_sensor_locations.csv",
    index=False
)
```
